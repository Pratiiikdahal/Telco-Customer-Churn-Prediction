### Encoding the data according to the column data type ie. Ordinal or Nominal

```python
from sklearn.model_selection import train_test_split

X=data.drop(columns=['Churn'],axis='columns')
Y=data['Churn']

X_train,X_test,Y_train,Y_test=train_test_split(X,Y,random_state=42,stratify=Y,test_size=0.2,shuffle=True)
```

### Label Encoding for the target variables 

```python
from sklearn.preprocessing import LabelEncoder

le=LabelEncoder()
y_train=le.fit_transform(Y_train)
y_test=le.transform(Y_test)

#inverse_transform can be used to turn the endoding back to the original labels
```

#encoding and standard scaling the columns here

```python
X_train.info()
```

```python
numerical_cols = X_train.select_dtypes(include=['int64', 'float64']).columns.tolist()
numerical_cols
```

```python
ordinal_columns=['Contract']
categorical_columns=X_train.select_dtypes( include=['str']).columns.tolist()#defining manually as there is just one
print(len(categorical_columns))
```

```python
categorical_columns=X_train.select_dtypes( include=['str']).columns.tolist()
ord_remove=ordinal_columns[0]
if ord_remove in categorical_columns:
    categorical_columns.remove(ord_remove)
len(categorical_columns)
print(categorical_columns)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import MinMaxScaler, OneHotEncoder, OrdinalEncoder

#defining ther order for the ordinal column
contract_order=[['Month-to-month', 'One year', 'Two year']]

feat_transform=ColumnTransformer(transformers=[('minmaxscaling',MinMaxScaler(),numerical_cols)
                                               ,('ordinalenc',OrdinalEncoder(categories=contract_order),ordinal_columns)
                                               ,('onehotenc',OneHotEncoder(handle_unknown='ignore'),categorical_columns)])
```

```python
X_train_transformed=feat_transform.fit_transform(X_train)
X_test_transformed = feat_transform.transform(X_test)
print(X_train_transformed.shape)
X_train_transformed[:5]
```

```python
!pip install imbalanced-learn
```

```python
from imblearn.over_sampling import SMOTE

sm=SMOTE(random_state=42)
x_train_resampled,y_train_resampled=sm.fit_resample(X_train_transformed,y_train)
```

```python
print(pd.Series(y_train_resampled).value_counts())
```

### As you can see after adding the Synthetic data using SMOTE the target column now looks balanced.

## 1.1 RandomForestClassifier

```python
#RandomForest
from sklearn.model_selection import RandomizedSearchCV
from sklearn.ensemble import RandomForestClassifier
from scipy.stats import randint

rf = RandomForestClassifier(random_state=42)
param_dist = {
    'n_estimators': randint(100, 500),
    'max_depth': [None, 10, 20, 30],
    'min_samples_split': randint(2, 10),
    'class_weight': ['balanced']
}
random_search = RandomizedSearchCV(
    rf,
    param_distributions=param_dist,
    n_iter=20, 
    cv=5,            
    scoring='f1',    
    n_jobs=-1,
    random_state=42
)
random_search.fit(x_train_resampled, y_train_resampled)
best_rf = random_search.best_estimator_
print("Best hyperparameters:", random_search.best_params_)
print("Best CV F1-score:", random_search.best_score_)
```

```python
ref_pred=best_rf.predict(X_test_transformed)
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test,ref_pred))
```

## 1.2 Decision Tree Classifier

```python
#Decision Trees
from sklearn.tree import DecisionTreeClassifier
treemodel=DecisionTreeClassifier(random_state=42)

parameters={
    'criterion':['gini','entropy'],
    'max_depth':[10,20,30,None],
    'class_weight': ['balanced']

}
randomized=RandomizedSearchCV(treemodel,param_distributions=parameters,cv=5,n_iter=50,scoring='accuracy',n_jobs=-1,random_state=42)


randomized.fit(x_train_resampled, y_train_resampled)

best_parameters=randomized.best_params_
print(best_parameters)

best_estimator=randomized.best_estimator_
print(best_estimator)
```

```python
dt_predicted=best_estimator.predict(X_test_transformed)
print(dt_predicted)
```

```python
print(classification_report(y_test, dt_predicted))
```

## 1.3 Logistic Regression

```python
from sklearn.model_selection import RandomizedSearchCV
from sklearn.linear_model import LogisticRegression
from scipy.stats import uniform

logreg = LogisticRegression(random_state=42, max_iter=5000)
param_dist = {
    'C': uniform(0.01, 10) ,
    'class_weight': ['balanced']     
}
random_search = RandomizedSearchCV(
    logreg, param_distributions=param_dist,n_iter=20, scoring='f1', cv=5,n_jobs=-1,random_state=42)
random_search.fit(x_train_resampled, y_train_resampled)
best_logreg = random_search.best_estimator_
print(random_search.best_params_)

y_pred_logreg = best_logreg.predict(X_test_transformed)

print(classification_report(y_test, y_pred_logreg))
```

## 1.4 SVM

```python
from sklearn.svm import SVC

svm = SVC(random_state=42)
param_dist = {
    'C': uniform(0.1, 10),            
    'kernel': ['linear', 'rbf', 'poly', 'sigmoid'],
    'gamma': ['scale', 'auto'],
    'class_weight': ['balanced']   
}

random_search = RandomizedSearchCV(
    svm,param_distributions=param_dist,n_iter=20,scoring='f1',cv=5,n_jobs=-1,random_state=42)

random_search.fit(x_train_resampled, y_train_resampled)
best_svm = random_search.best_estimator_
print(random_search.best_params_)
```

```python
y_pred_svm = best_svm.predict(X_test_transformed)
```

```python
print(classification_report(y_test, y_pred_svm))
```
